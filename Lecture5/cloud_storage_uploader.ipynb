```python
from abc import ABC, abstractmethod
class StorageClient:
  def __init__(self, object_name, data):
    self.object_name = object_name
    self.data = data
  
  @abstractmethod
  def upload(self, path, data):
    pass
```

```python
class LocalStorage(StorageClient):
  def __init__(self, root='/'):
    self.root = root

  def upload(self, path, data):
    print(f'Request Accepted')
    print(f'{data} is being uploaded to {self.root + path}')
    print(f'Data saved to local storage successfully!')
```

```python
class S3Bucket(StorageClient):
  def __init__(self, bucket_name='danish-s3'):
    self.bucket_name = bucket_name

  def upload(self, path, data):
    print(f'Request Accepted')
    print(f'{data} is being uploaded to s3 bucket {self.bucket_name} at {path}')
    print(f'Data saved to s3 bucket successfully!')
```

```python
class GCS(StorageClient):
  def __init__(self, bucket_name='danish-gcp'):
    self.bucket_name = bucket_name

  def upload(self, path, data):
    print(f'Request Accepted')
    print(f'{data} is being uploaded to Google Cloud Storage {self.bucket_name} at {path}')
    print(f'Data saved to GCS bucket successfully!')
```

```python
lfs = LocalStorage()
aws_s3 = S3Bucket()
gcs = GCS()
```

```python
lfs.upload(data='index.py', path='lecture_5')
```

```python
aws_s3.upload(data='index.py', path='lecture_5')
```

```python
gcs.upload(data='index.py', path='lecture_5')
```

```python
# For Loop Implementation
storages = [
  LocalStorage(),
  S3Bucket(),
  GCS()
]
data = 'lecture_5_notes.txt'
path = 'enpm818q/python/'

for storage in storages:
  storage.upload(path, data)
```

```python
# Factory Creation
class StorageFactory():
  @staticmethod
  def create(kind, **kwargs):
    kind = kind.lower()
    if kind == 'local':
      return LocalStorage(**kwargs)
    if kind == 's3':
      return S3Bucket(**kwargs)
    if kind == 'gcs':
      return GCS(**kwargs)
    
    raise ValueError(f"Unknown storage kind")
```

```python
#Singleton

class ConfigManager:
  _instance = None

  def __new__(cls):
    if cls._instance is None:
      cls._instance = super().__new__(cls)
      cls._instance.setting = {}
      print('Initialising ConfigManager')
    return cls._instance
```

```python
cfg1 = ConfigManager()
cfg2 = ConfigManager()

cfg1.setting["mode"] = 'DEV'

print(cfg1.setting["mode"])
print(cfg2.setting["mode"])
print(cfg1 is cfg2)
print(cfg1)
print(cfg2)
```
