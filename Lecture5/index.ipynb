```python
class CloudResource:
  def __init__(self):
    print('Hi, Cloud Resource')

  def provision(self):
    print('Hi, Cloud Resource is being provisioned.')
```

```python
class ComputerResource(CloudResource):
  pass
```

```python
vm = CloudResource()
vm.provision()
```

```python
class CloudResource:
  def __init__(self, name):
    self.name = name

class ComputeResource:
  def __init__(self, name):
    self.name = name
```

```python
from abc import ABC, abstractmethod

class CloudResource(ABC):
  def __init__(self, name):
    self.name = name

  def log(self, message):
    print(f'{self.name}: {message}')

  @abstractmethod
  def provision(self):
    pass
```

```python
class StorageBucket(CloudResource):
  pass
```

```python
storage_bucket = StorageBucket()
```
