```python
class Payment:
  def __init__(self, amount, currency):
    self.amount = amount
    self.currency = currency
  
  def process_payment(self):
    print('Generic Payment Processing')
```

```python
class CreditCardPayment(Payment):
  def __init__(self, amount, currency, card_number):
    super().__init__(amount, currency)
    self.card_number = card_number

  def process_payment(self):
    print(f'Payment {self.amount}/{self.currency} is being processed through Credit Card *****{self.card_number[-4:]}')
```

```python
class PayPalPayment(Payment):
  def __init__(self, amount, currency, user_name):
    super().__init__(amount, currency)
    self.user_name = user_name

  def process_payment(self):
    print(f'Redirecting to PayPal.....')
    print(f'Logging {self.user_name}.....')
    print(f'Authenticating {self.user_name}.....')
    print(f'Authorising {self.user_name}.....')
    print(f'Payment {self.amount}/{self.currency} is being processed through PayPal')
    print(f'Payment Processed for {self.user_name}. Thanks!')
```

```python
class CryptoPayment(Payment):
  def __init__(self, amount, currency, wallet_id):
    super().__init__(amount, currency)
    self.wallet_id = wallet_id

  def process_payment(self):
    print(f'Redirecting to Binance.....')
    print(f'Payment {self.amount}/{self.currency} is being processed through Binance')
    print(f'Payment Processed by {self.wallet_id}. Thanks!')
```

```python
payments = [
  CryptoPayment(123, 'BTC', 'dajksn8273647jnsjkfabs'),
  PayPalPayment(20, '$', 'danitrical'),
  CreditCardPayment(3154, 'PKR', '4242424242424242')
]
```

```python
for p in payments:
  p.process_payment()
```
