```python
import datetime as dt
class Logger:
  _instance = None

  def __new__(cls):
    if cls._instance is None:
      cls._instance = super().__new__(cls)
      cls._instance.time = dt.time
      cls._instance.date = dt.date
    return cls._instance
  
  def __init__(self, logs=[]):
    self.logs = logs

  def log(self, message):
    timestamp = dt.datetime.now().strftime("%H:%M:%S")
    self.logs.append(f'[{timestamp}]: {message}')

  def show_logs(self):
    for msg in self.logs:
      print(msg)
```

```python
logger_1 = Logger()
logger_2 = Logger()
```

```python
logger_1.log('App initiating')
logger_2.log('App started')
logger_1.log('App running')
logger_2.log('Phrrr Phrrrr Phrrrrr')
```

```python
logger_1.show_logs()
```

```python
logger_2.show_logs()
```
