```python
# Method Overriding
```

```python
class Notification:
  def __init__(self, recipient, message):
    self.recipient = recipient
    self.message = message

  def prepare_message(self):
    print(f'Started sending message to {self.recipient}')

  def finalise(self):
    print(f'Done sending message to {self.recipient}')

  def send(self):
    print(f'Sending message: {self.message} to {self.recipient}')
```

```python
class EmailNotification(Notification):
  def __init__(self, recipient, message):
    # self.recipient = recipient
    # self.message = message
    super().__init__(recipient, message)


  def send(self):
    self.prepare_message()
    print(f'(Email) {self.message} sent to {self.recipient}')
    self.finalise()
```

```python
class SMSNotification(Notification):
  def __init__(self, recipient, message):
    # self.recipient = recipient
    # self.message = message
    super().__init__(recipient, message)


  def send(self):    
    self.prepare_message()
    print(f'(SMS) {self.message} sent to {self.recipient}')
    self.finalise()
```

```python
class PushNotification(Notification):
  def __init__(self, recipient, message):
    # self.recipient = recipient
    # self.message = message
    super().__init__(recipient, message)


  def send(self):
    self.prepare_message()
    print(f'(Push Notification) {self.message} sent to {self.recipient}')
    self.finalise()
```

```python
email = EmailNotification('danitric@umd.edu', 'Hi, You got a job offer')
sms = SMSNotification('+17347443268', 'Hi, You got a job offer')
push = PushNotification('Android App', 'Hi, You got a job offer')
```

```python
email.send()
```
